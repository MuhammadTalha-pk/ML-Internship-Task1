## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Research question:
Can observable search, content, and engagement signals from the previous 90 days help identify which pages are at higher risk of meaningful decline during the following 30 days?
Unit of analysis:
One content page at a defined prediction date.
Decision:
Which pages should a content or SEO reviewer inspect first because they may be at higher risk of future decline?
Output:
A ranked list of pages with a future-decline risk score and supporting reason codes.
Action:
A reviewer can inspect the highest-risk pages first and decide whether to monitor, update, improve, protect, or leave them unchanged.
Cost of a wrong recommendation:
A false positive may waste reviewer time on a page that would have remained stable. A false negative may cause the team to miss a page that later declines and should have been reviewed earlier.
Why data or ML may help:
Page performance depends on several signals such as impressions, clicks, CTR, average position, content age, sessions, and engagement. A simple hand-written rule may miss combinations and changes across these signals. ML may help if these relationships are real but too complex to capture with one rule.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
unique_clients = df["client_id"].nunique()

current_down_rate = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .mean()
)

corr = df["search_volume"].corr(df["impressions_90d"])

print("Total pages:", total_pages)
print("Unique clients:", unique_clients)
print("Current 'down' rate:", round(current_down_rate * 100, 1), "%")
print("Search volume vs impressions correlation:", round(corr, 3))
```
